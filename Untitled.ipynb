```python
import pandas as pd

iris_array = [{'sepal_lenght': 'NaN', 'sepal_wight': 'NaN', 'petal_lenght': 'NaN', 'petal_wight': 0.2, 'encoded_target': 0 },
            {'sepal_lenght': 'NaN', 'sepal_wight': 3.0, 'petal_lenght': 1.4, 'petal_wight': 0.2, 'encoded_target': 0 },
            {'sepal_lenght': 5.0, 'sepal_wight': 3.6, 'petal_lenght': 1.4, 'petal_wight': 'NaN', 'encoded_target': 0 },
            {'sepal_lenght': 10.0, 'sepal_wight': 3.0, 'petal_lenght': 1.0, 'petal_wight': 0.3, 'encoded_target': 0 }]


iris_df = pd.DataFrame(iris_array)
iris_df.to_csv('./data/null_array.csv', index=False) # index=False remove "Unnamed: 0".
file_iris_df = pd.read_csv('./data/null_array.csv') # drop "Unnamed: 0" column when is reading the file with parameter ", index_col=0"
#localizando especificamente linhas com qualquer coluna nula
file_iris_df[file_iris_df.isna().any(axis=1)]

# lidando com dados nulos, substituição

# 1. opcao - substituir pela média da coluna
iris_df_mean = file_iris_df.fillna(file_iris_df.mean())

# 2. Substituir por zero
iris_df_zero = file_iris_df.fillna(0)

# 3. Substituir por um valor impossivel
iris_df_valores_impossiveis = file_iris_df.fillna(-999)

print("Mean:")
display(iris_df_mean)

print("Zero:")
display(iris_df_zero)

print("Impossible value (-999):")
display(iris_df_valores_impossiveis)
```

```python
import pandas as pd
#range(10, 0, -2) means: start at 10, stop before 0, step by -2.
#It counts down by 2 each time and never reaches 0, because the stop value is excluded.
#It produces 10, 8, 6, 4, 2.
#So 'B': range(10, 0, -2) creates a column named B with those values.
#The three arguments are range(start, stop, step):
#start: where it begins (10)
#stop: where it ends, not included (0)
#step: how much to add each time (-2, so it goes down)

#range(1, 6) means: start at 1, stop before 6, step by 1. 
#The step is optional and defaults to 1, so it counts up by one and never reaches 6, 
#because the stop value is excluded.
#It produces 1, 2, 3, 4, 5.

#A: range(1, 6) gives 1, 2, 3, 4, 5
#B: range(10, 0, -2) gives 10, 8, 6, 4, 2
#C: range(10, 5, -1) gives 10, 9, 8, 7, 6


df = pd.DataFrame({'A': range(1, 6),'B': range(10, 0, -2),'C': range(10, 5, -1)})
print(df.query('A > B'))
print('------------------')
print(df.query('B == `C`'))
print('------------------')
print(df[df.A > df.B])




 
 
```

```python
import pandas as pd

df1 = pd.DataFrame({'A': range(1, 6),'B': range(10, 0, -2),'C': range(10, 5, -1)})
drop_column_df = df1.drop(columns=['A'])
print('Drop Column ------------------')
print(drop_column_df)
 
df2 = pd.DataFrame({'A': range(1, 6),'B': range(10, 0, -2),'C': range(10, 5, -1)})
drop_row_df = df2.drop([0,1])
print('Drop Row ------------------')
print(drop_row_df)

df3 = pd.DataFrame({'A': range(1, 6),'B': range(10, 0, -2),'C': range(10, 5, -1)})
drop_column_df3 = df3.drop(columns=['A'])
drop_column_and_row_df3 = drop_column_df3.drop([0,1])

print('Drop Clumn A and 0 and 1 Row ------------------')
print(drop_column_and_row_df3)

drop_column_and_row_undex_reseted_df3 = drop_column_and_row_df3.reset_index(drop=True)
print('Drop Clumn A and 0 and 1 Row Index Reseted ------------------')
print(drop_column_and_row_undex_reseted_df3)
```

```python
#Contat example
import pandas as pd


s1 = pd.Series(['a', 'b'])
s2 = pd.Series(['c', 'd'])
pd.concat([s1, s2], axis=1)

#merge example

df1 = pd.DataFrame({'lkey': ['foo', 'bar', 'baz', 'foo'],
                   'value': [1, 2, 3, 5]})
df2 = pd.DataFrame({'rkey': ['foo', 'bar', 'baz', 'foo'],
                    'value': [5, 6, 7, 8]})
print(df1.merge(df2, left_on='lkey', right_on='rkey', suffixes=("_left","_right")))

#Join Example
df = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3', 'K4', 'K5'],
                    'A': ['A0', 'A1', 'A2', 'A3', 'A4', 'A5']})
other = pd.DataFrame({'key': ['K0', 'K1', 'K2'],
                       'B': ['B0', 'B1', 'B2']})
df.join(other, suffixes=("_caller","_other"))
 
 
```
