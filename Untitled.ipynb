```python
class cachorro:
    def __init__(self, nome, raca, tutor):
        self.nome = nome
        self.raca = raca
        self.tutor = tutor

    def latir(self):
        return "au! au!!"
```

```python
terry = cachorro("Terry", "Pastor Alemão", "Fernando")

print("Dados do Terry:")
print("Nome:", terry.nome)
print("Raça:", terry.raca)
print("Tutor:", terry.tutor)
print("------------")
print(terry.latir())
```
