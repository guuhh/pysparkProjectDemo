```python
import pandas as pd
from sklearn.datasets import load_iris
import plotly.express as px

#data = load_iris()
#iris_df = pd.DataFrame(data.data, columns = data.feature_names)
#iris_df.columns =  [c.replace("(cm)","").rstrip().replace(" ","_") for c in iris_df.columns]
#display(iris_df)
iris_df = px.data.iris()
iris_df.columns =  [c.replace("(cm)","").rstrip().replace(" ","_") for c in iris_df.columns]
display(iris_df.columns)   # shows all column names
display(iris_df.head())    # shows the first 5 rows

# Each flower as a point
fig1 = px.scatter(iris_df, x="sepal_length", y="species_id")
fig1.show()

# Or, more readable, one box per species
fig2 = px.box(iris_df, x="species", y="sepal_length")
fig2.show()

# Or the histogram colored by species
fig3 = px.histogram(iris_df, x="sepal_length", color="species")
fig3.show()
```
